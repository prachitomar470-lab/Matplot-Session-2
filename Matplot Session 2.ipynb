```python

import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
steps = [2000, 3000, 4000, 3500, 5000, 6000, 7000]

plt.plot(days, steps, color="green", linewidth=2)
plt.show()

```

```python

import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5, 6]
users = [100, 150, 200, 250, 300, 350]

plt.plot(months, users, marker="o", markerfacecolor="red")
plt.show()

```

```python

import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
zomato = [5, 8, 6, 10, 12]
swiggy = [4, 6, 9, 7, 11]

plt.plot(months, zomato, color="red", linestyle="-", label="Zomato")
plt.plot(months, swiggy, color="blue", linestyle="--", label="Swiggy")

plt.legend()
plt.show()

```

```python

import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5, 6]
data = [2, 4, 3, 5, 6, 8]

plt.plot(months, data)
plt.title("Monthly Data Usage Trend")
plt.xlabel("Month")
plt.ylabel("Data Used (GB)")

plt.show()

```
